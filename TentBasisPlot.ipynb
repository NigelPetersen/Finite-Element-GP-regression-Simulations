```
import numpy as np
import matplotlib.pyplot as plt


J = 4
h = 1 / J
x = np.linspace(0, 1, 1000)
plt.figure(figsize=(10, 6))


psi_0 = np.zeros_like(x)
mask0 = (x >= 0) & (x <= h)
psi_0[mask0] = -x[mask0] / h + 1

plt.plot(x, psi_0, linewidth=2, label=r"$e_{h,0}(x)$")


for i in range(1, J):

    psi = np.zeros_like(x)
    mask1 = (x >= (i - 1) * h) & (x <= i * h)
    psi[mask1] = x[mask1] / h - i + 1
    mask2 = (x >= i * h) & (x <= (i + 1) * h)
    psi[mask2] = -x[mask2] / h + i + 1

    plt.plot(x, psi, linewidth=2, label=rf"$e_{{h,{i}}}(x)$")


psi_K = np.zeros_like(x)
maskK = (x >= (J - 1) * h) & (x <= J * h)
psi_K[maskK] = x[maskK] / h - J + 1

plt.plot(x, psi_K, linewidth=2, label=r"$e_{{h,{K}}}(x)$")


grid = np.arange(J + 1) * h
plt.axhline(0, linewidth=0.8)
plt.ylim(0.0, 1.2)
plt.xlim(0, 1)
plt.grid(alpha=0.3)
plt.legend(loc = "upper right", ncol = 5)

plt.show()
```
